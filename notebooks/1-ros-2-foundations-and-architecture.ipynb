# ROS 2 Foundations and Architecture

This required concept module comes before Duckiedrone data paths and Robot Operating System 2 (ROS 2) commands. ROS 2 is a set of software libraries and tools for building distributed robotic applications. It is not a Linux distribution or a replacement for the operating system. No ROS 2 installation, command, or Duckiedrone is required for this module.

## The ROS 2 graph

ROS 2 organizes a robotic application as a graph of independent nodes. A node has a focused responsibility, such as reading a sensor, estimating state, planning a path, or controlling an actuator. Nodes exchange typed data through defined interfaces rather than directly sharing implementation details.

```text
sensor node -> typed topic -> perception node -> typed topic -> control node
```

Topics carry streams of messages between publishers and subscribers. Services model a request and response. Actions represent longer-running work with a goal, feedback, a result, and cancellation. Parameters hold configuration values for a node. These are graph interfaces, not separate operating-system features.

## Software layers

At a simplified level, an application node uses a language client library such as `rclpy` for Python or `rclcpp` for C++. Those libraries use common ROS client-library layers and the ROS Middleware (RMW) interface, exposed to code as `rmw`. The selected middleware implementation handles the communication and discovery details beneath the ROS 2 graph Application Programming Interface (API).

```text
application node
  |
rclpy or rclcpp client library
  |
ROS client library and RMW interface
  |
selected middleware implementation, discovery, and transport
```

This separation lets the same ROS 2 application programming model work with supported middleware implementations. Do not assume that every ROS 2 system uses the same middleware or discovery configuration. The Duckiedrone-specific middleware, domain, and naming context is introduced later, after this model is clear.

## ROS 2 is distributed

Nodes can run in one process, in several processes on one computer, or across authorized computers that share compatible configuration. The `ros2` command-line tools inspect and interact with the graph visible to the shell that runs them. A local terminal does not automatically see a graph running on a Duckiedrone, and a physical-device graph must not be reconfigured merely to make it visible.

The next module applies this general model to Duckiedrone sensor and flight-control data paths. Later modules separately cover interfaces, discovery and domains, local package development, and authorized read-only device inspection.
